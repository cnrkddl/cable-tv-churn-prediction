그냥 반복문 만들어서 한번에 모든 칼럼들의 구성 데이터와 결측치 여부 등을 한번에 알아보자 ! 

```python
cols = df_tps.columns.tolist()

for i in range(0, 10):
    col = cols[i]

    print(f"[{col}]")
    print(df_tps[col].value_counts(dropna=False))
    print("-------------")

    # 공통
    nan_cnt = df_tps[col].isna().sum()

    # 숫자형 컬럼 처리
    if pd.api.types.is_numeric_dtype(df_tps[col]):
        zero_cnt     = (df_tps[col] == 0).sum()
        non_zero_cnt = (df_tps[col] != 0).sum()

        print(f"zero_cnt     : {zero_cnt}")
        print(f"nan_cnt      : {nan_cnt}")
        print(f"non_zero_cnt : {non_zero_cnt}")

    # 문자열 컬럼 처리
    else:
        empty_cnt = df_tps[col].astype(str).str.strip().eq('').sum()
        space_cnt = df_tps[col].astype(str).str.match(r'^\s+$').sum()

        print(f"empty_string_cnt : {empty_cnt}")   # '' 또는 공백 제거 후 빈 값
        print(f"space_only_cnt   : {space_cnt}")   # '   ' 같은 순수 공백
        print(f"nan_cnt          : {nan_cnt}")

    print("\n")
```
